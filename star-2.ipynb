```python
def invalid(e):
    if ((len(e)%2)==1) :
        return False
    else :
        if (e[:int(len(e)/2)]==e[int(len(e)/2):]) :
            return True
        else :
            return False
```

```python
def parcourir(l):
    x=0
    for i in range(int(l[0]),int(l[1])+1) :
        if (invalid(str(i))) :
            #print(i)
            x+=i
    return x
        
```
